```python
from pathlib import Path
from ultralytics import YOLO

ROOT = Path('/root/hit-uav')
DATA_YAML = ROOT / 'dataset.yaml'

model = YOLO('yolov8n.pt')  

# ============ 3. 微调训练 ============
results = model.train(
    data=str(DATA_YAML),
    epochs=100,             
    imgsz=640,              
    batch=16,    #实际有效batch=64,           
    workers=8,
    device=0,               
    seed=0,                
    patience=30,            
    # ---- 优化器与学习率 ----
    optimizer='auto',         #实际生效AdamW 
    lr0=0.01,                 # 无效，实际0.00125，被覆盖
    lrf=0.01,               
    momentum=0.937,              #实际0.9
    weight_decay=0.0005,
    warmup_epochs=3.0,
    cos_lr=False, # 余弦退火

    # ---- 预训练与冻结策略 ----
    pretrained=True,       
    freeze=None,           
                                           
    # ---- 数据增强（红外域重点）----
    hsv_h=0.0,
    hsv_s=0.0,
    hsv_v=0.0,
    degrees=0.0,              # 航拍红外一般不做旋转
    translate=0.0,            # 轻微平移
    scale=0.0,                # 尺度抖动
    fliplr=0.0,              # 左右翻转
    flipud=0.0,
    mosaic=0.0,               # Mosaic 
    mixup=0.0,  # 红外图 mixup 收益小，关闭
    close_mosaic=0,          

    # ---- 验证与输出 ----
    val=True,
    plots=True,
    name='yolov8n_finetune',   
    project=str(ROOT / 'yolov8n_results'),
)

# ============ 4. 训练后最终验证 ============
metrics = model.val()
print(f"验证集 mAP50 = {metrics.box.map50:.4f}   mAP50-95 = {metrics.box.map:.4f}")
```

```python
from pathlib import Path
import json
import numpy as np
from ultralytics import YOLO

DATA_ROOT = Path("/root/hit-uav")
IMG_DIR = DATA_ROOT / "images" / "test"
OUT_DIR = DATA_ROOT / "yolov8n_results" / "true_predictions"
OUT_DIR.mkdir(parents=True, exist_ok=True)

MODELS = {
    "yolov8n":  "/root/hit-uav/yolov8n_results/yolov8n_finetune/weights/best.pt",
}

files = sorted(IMG_DIR.glob("*.jpg"))
print(f"测试集图片数: {len(files)}")

for name, weight in MODELS.items():
    print(f"\n推理 {name} ...")
    model = YOLO(weight)
    pred_by_img = {}

    for f in files:
        r = model.predict(
            source=str(f),
            conf=0.001,       
            iou=0.7,           # NMS 阈值统一
            max_det=300,
            imgsz=640,
            verbose=False,
        )[0]

        dets = []
        if r.boxes is not None:
            for box, cls, conf_bit in zip(
                r.boxes.xyxy.tolist(),
                r.boxes.cls.tolist(),
                r.boxes.conf.tolist()
            ):
                dets.append([
                    int(cls),                      # HIT-UAV 类别索引
                    round(box[0], 2), round(box[1], 2),
                    round(box[2], 2), round(box[3], 2),
                    round(float(conf_bit), 6),
                ])
        pred_by_img[f.stem] = dets

    out_path = OUT_DIR / f"{name}_pred.json"
    json.dump(pred_by_img, open(out_path, "w"), ensure_ascii=False, indent=2)
    total = sum(len(v) for v in pred_by_img.values())
    print(f"{name}: {len(pred_by_img)} 张，总框数 {total}")
    print(f"保存: {out_path}")
```
