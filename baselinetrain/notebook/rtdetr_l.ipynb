```python
from pathlib import Path
from ultralytics import RTDETR

ROOT = Path('/root/hit-uav')
DATA_YAML = ROOT / 'dataset.yaml'   

model = RTDETR('rtdetr-L.pt')   # COCO 预训练

results = model.train(
    data=str(DATA_YAML),
    epochs=120,   #收敛较慢，比YOLO多20           
    imgsz=640,
    batch=8,         #YOLO是16，但不影响对比，实际都是64       
    workers=8,
    device=0,
    seed=0,
    patience=40,      #架构不同，比YOLO高10       

    optimizer='AdamW',
    lr0=1e-4,                # 比 YOLO 小 100 倍
    lrf=0.01,
    weight_decay=1e-4,       #YOLO是0.0005
    warmup_epochs=3,

    # ---- 预训练 ----
    pretrained=True,
    freeze=None,             

    # ---- 增强：红外无颜色，几何增强保留 ----
    hsv_h=0.0, hsv_s=0.0, hsv_v=0.0,
    degrees=0.0,
    translate=0.0,
    scale=0.0,               # 小目标很重要
    fliplr=0.0,
    flipud=0.0,
    mosaic=0.0,             
    mixup=0.0,
    close_mosaic=0,

    val=True,
    plots=True,
    name='rtdetr_l_finetune',
    project=str(ROOT / 'rtdetr_l_results'),
)

metrics = model.val()
print(f"验证集 mAP50 = {metrics.box.map50:.4f}   mAP50-95 = {metrics.box.map:.4f}")
```

```python
from pathlib import Path
import json
import numpy as np
from ultralytics import RTDETR 

DATA_ROOT = Path("/root/hit-uav")
IMG_DIR = DATA_ROOT / "images" / "test"
OUT_DIR = DATA_ROOT / "rtdetr_r50_results" / "true_predictions"
OUT_DIR.mkdir(parents=True, exist_ok=True)

MODELS = {
   "rtdetr_r50": "/root/hit-uav/rtdetr_r50_results/rtdetr_r50_finetune/weights/best.pt", 
}

files = sorted(IMG_DIR.glob("*.jpg"))
print(f"测试集图片数: {len(files)}")

for name, weight in MODELS.items():
    print(f"\n推理 {name} ...")
    model = RTDETR(weight)
    pred_by_img = {}

    for f in files:
        r = model.predict(
            source=str(f),
            conf=0.001,       
            iou=0.7,           
            max_det=300,
            imgsz=640,
            verbose=False,
        )[0]

        dets = []
        if r.boxes is not None:
            for box, cls, conf_bit in zip(
                r.boxes.xyxy.tolist(),
                r.boxes.cls.tolist(),
                r.boxes.conf.tolist()
            ):
                dets.append([
                    int(cls),                      # HIT-UAV 类别索引
                    round(box[0], 2), round(box[1], 2),
                    round(box[2], 2), round(box[3], 2),
                    round(float(conf_bit), 6),
                ])
        pred_by_img[f.stem] = dets

    out_path = OUT_DIR / f"{name}_pred.json"
    json.dump(pred_by_img, open(out_path, "w"), ensure_ascii=False, indent=2)
    total = sum(len(v) for v in pred_by_img.values())
    print(f"{name}: {len(pred_by_img)} 张，总框数 {total}")
    print(f"保存: {out_path}")
```
